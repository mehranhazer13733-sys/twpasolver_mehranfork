## Basic 4WM Analysis for different pump frequencies

```python
# ============================================================
# BASIC 4WM TWPASOLVER NOTEBOOK BLOCK
# Latest-version twpasolver assumed
# ============================================================

%matplotlib inline
import twpasolver

import numpy as np
import matplotlib.pyplot as plt

from twpasolver.models import TWPA, LCLfBaseCell
from twpasolver import TWPAnalysis
from twpasolver import ModeSet
from twpasolver.cmes.solver_config import BlochSolverConfig


# ============================================================
# 1. USER INPUTS
# ============================================================

# ============================================================
# DISTRIBUTED LOSS SETTINGS
# ============================================================

DIELECTRIC_LOSS_TANGENT =0
QP_LOSS_PER_HZ = 0

# Frequency grid for linear response
# Use 50 GHz if you later want higher products like 3fp or 4fp-fs.
freqs_hz = np.arange(1e9, 50e9, 2e6)
freqs_ghz = freqs_hz / 1e9

# Geometry
p_cell = 2.21e-6              # unit-cell pitch [m]
N_stubs_sc = 59               # cells per modulation period / supercell
N_supercells = 920            # number of repeated supercells. 825 for 100mm design. 385 for 50mm design
P_mod = N_stubs_sc * p_cell   # modulation period [m]

device_length_mm = N_stubs_sc * p_cell * N_supercells * 1e3

# Film / inductance
L_cell = 32e-12             # H per main-line cell # from geometry + film thickness
Lf_stub = 0.0                 # H, stub inductance for LCLfBaseCell

# Capacitance model
# cprime = little c' = capacitance per length of one strip/stub section
cprime = 0.8581e-9/2             # F/m # from ads sim

# Length modulation
l_avg = 10.8e-6               # average stub length [m]
l_mod = 1.04e-6               # peak-to-peak length modulation [m]

# Nonlinear KI parameters
Istar = 3.2e-3                # A, design nonlinear-current scale
Idc = 0.0                     # A, 4WM case

# Reference impedance for S-parameters and power <-> current conversion
Z0_ref = 50.0

# ============================================================
# 2. WORKFLOW SETTINGS -- EDIT SETTINGS HERE
# ============================================================
# All gain-producing cells use exactly this model, signal grid, spatial
# output sampling, analysis range, and Bloch solver configuration. Pump-axis
# and power-axis increments remain block-specific because they select
# operating points; they do not change the resolution of a gain trace.
GAIN_MODEL = "bloch"
GAIN_SIGNAL_MIN_GHZ = 1.0
GAIN_SIGNAL_MAX_GHZ = 18.0
GAIN_FREQUENCY_STEP_GHZ = 0.025
GAIN_SPATIAL_THIN = 100
ANALYSIS_FREQUENCY_RANGE_GHZ = (0.005, 50.0, 0.005)
PUMP_STOPBAND_GUARD_GHZ = 0.70
PUMP_CURRENT_WARNING_RATIO = 0.30

BLOCH_CONFIG = BlochSolverConfig(
    passes=80,
    anderson_depth=30,
    convergence_threshold=0.01,
)

GAIN_SIGNAL_POINT_COUNT = int(
    round((GAIN_SIGNAL_MAX_GHZ - GAIN_SIGNAL_MIN_GHZ) / GAIN_FREQUENCY_STEP_GHZ)
) + 1
GAIN_SIGNAL_FREQS_GHZ = np.linspace(
    GAIN_SIGNAL_MIN_GHZ, GAIN_SIGNAL_MAX_GHZ, GAIN_SIGNAL_POINT_COUNT
)

# Main selected-pump plot and current profile.
MAIN_PUMP_POWER_DBM = -25.0
MAIN_SIGNAL_POWER_DBM = -80.0
SELECTED_PUMP_CANDIDATES_GHZ = np.array(
    [10.0, 10.75, 11.8, 12.7], dtype=float
)
CURRENT_PROFILE_PUMP_GHZ = 12.7
CURRENT_PROFILE_SIGNAL_GHZ = 6.0

# Phase-matching heatmap and diagnostic plots.
PHASE_MATCH_SIGNAL_MIN_GHZ = 4.0
PHASE_MATCH_SIGNAL_MAX_GHZ = 8.0
PHASE_MATCH_PUMP_MIN_GHZ = 9.0
PHASE_MATCH_PUMP_MAX_GHZ = 14.5
PHASE_SLICE_PUMP_GHZ = 13.8
PHASE_DIAGNOSTIC_PUMP_GHZ = 10.6
RAW_K_DIAGNOSTIC_MIN_GHZ = 0.1
RAW_K_DIAGNOSTIC_MAX_GHZ = 24.0
RAW_K_DIAGNOSTIC_POINTS = 4000
BETA_DIAGNOSTIC_MIN_GHZ = 2.0
BETA_DIAGNOSTIC_MAX_GHZ = 20.0
BETA_DIAGNOSTIC_POINTS = 4000
PLOT_NEGATIVE_MISMATCH = True
BETA_CHECK_SIGNAL_FREQS_GHZ = np.array([4.0, 5.0, 6.0, 7.0, 8.0])

# Pump-frequency / pump-power heatmap and its starred final curve.
HEATMAP_PUMP_POWERS_DBM = np.arange(-23.0, -19.0, 0.50)
HEATMAP_PUMP_MIN_GHZ = 10.0
HEATMAP_PUMP_MAX_GHZ = 12.7
HEATMAP_PUMP_STEP_GHZ = 0.50
HEATMAP_SIGNAL_POWER_DBM = -80.0
HEATMAP_EVALUATION_MIN_GHZ = 4.0
HEATMAP_EVALUATION_MAX_GHZ = 8.0
HEATMAP_MAX_PUMP_CURRENT_RATIO = 0.20
if not (
    GAIN_SIGNAL_MIN_GHZ <= HEATMAP_EVALUATION_MIN_GHZ
    < HEATMAP_EVALUATION_MAX_GHZ <= GAIN_SIGNAL_MAX_GHZ
):
    raise ValueError("The heatmap evaluation band must lie inside the gain grid.")

# Pump-power sweep.
PUMP_POWER_SWEEP_N_SUPERCELLS = 920
PUMP_POWER_SWEEP_PUMP_GHZ = 12.7
PUMP_POWER_SWEEP_POWERS_DBM = np.array(
    [-25.0, -24.0, -23.0, -21.0, -20.0], dtype=float
)
PUMP_POWER_SWEEP_SIGNAL_POWER_DBM = -80.0

# Signal-power sweep.
SIGNAL_POWER_SWEEP_N_SUPERCELLS = 920
SIGNAL_POWER_SWEEP_PUMP_GHZ = 10.8
SIGNAL_POWER_SWEEP_PUMP_POWER_DBM = -21.0
SIGNAL_POWER_SWEEP_POWERS_DBM = np.arange(-110.0, -59.0, 10.0)

# Device-length sweep.
LENGTH_SWEEP_N_SUPERCELLS = np.array(
    [100, 200, 300, 385, 500, 650, 825, 920, 1000], dtype=int
)
LENGTH_SWEEP_PUMP_GHZ = 10.6
LENGTH_SWEEP_PUMP_POWER_DBM = -24.0
LENGTH_SWEEP_SIGNAL_POWER_DBM = -90.0


def dBm_to_Ipeak(P_dBm, Z0=50.0):
    """
    Convert available power in dBm into sinusoidal peak current.
    Assumes P = 0.5 * I_peak^2 * Z0.
    """
    P_W = 1e-3 * 10**(P_dBm / 10)
    I_peak = np.sqrt(2 * P_W / Z0)
    return I_peak


def Ipeak_to_dBm(I_peak, Z0=50.0):
    """
    Convert sinusoidal peak current into dBm.
    Assumes P = 0.5 * I_peak^2 * Z0.
    """
    P_W = 0.5 * I_peak**2 * Z0
    return 10 * np.log10(P_W / 1e-3)


def inclusive_frequency_grid(start_ghz, stop_ghz, step_ghz):
    """Return a stable grid whose first and last values are exact."""
    start_ghz = float(start_ghz)
    stop_ghz = float(stop_ghz)
    step_ghz = float(step_ghz)
    n_steps = int(round((stop_ghz - start_ghz) / step_ghz))
    if not np.isclose(
        start_ghz + n_steps * step_ghz, stop_ghz, rtol=0.0, atol=1e-12
    ):
        raise ValueError("The requested interval must be an integer number of steps.")
    return np.linspace(start_ghz, stop_ghz, n_steps + 1)


def pump_is_stopband_safe(analysis_obj, pump_ghz, guard_ghz=PUMP_STOPBAND_GUARD_GHZ):
    """Reject pumps in or near a linear stopband before an expensive solve."""
    for lower, upper in np.asarray(analysis_obj.linear.stopbands, dtype=float):
        if lower - guard_ghz <= pump_ghz <= upper + guard_ghz:
            return False
    return True


def valid_4wm_signal_mask(analysis_obj, pump_ghz, signal_freqs_ghz):
    """Physical basic-4WM points, excluding only the exact p=s=i degeneracy."""
    signal_freqs_ghz = np.asarray(signal_freqs_ghz, dtype=float)
    idler_freqs_ghz = 2.0 * float(pump_ghz) - signal_freqs_ghz
    sorted_unique = np.unique(signal_freqs_ghz)
    positive_steps = np.diff(sorted_unique)
    positive_steps = positive_steps[positive_steps > 0]
    degeneracy_tolerance = (
        0.5 * np.min(positive_steps) if len(positive_steps) else 1e-9
    )
    f_min = float(np.min(analysis_obj.freqs))
    f_max = float(np.max(analysis_obj.freqs))
    valid = (
        (signal_freqs_ghz >= f_min)
        & (signal_freqs_ghz <= f_max)
        & (idler_freqs_ghz >= f_min)
        & (idler_freqs_ghz <= f_max)
        & (np.abs(signal_freqs_ghz - pump_ghz) > degeneracy_tolerance)
    )
    return valid, idler_freqs_ghz


def run_gain_curve(
    analysis_obj,
    Is0_value,
    pump_ghz,
    *,
    Ip0_value=None,
    modes="basic_4wm",
    enforce_pump_guard=True,
):
    """Run one gain curve with the shared grid and Bloch configuration."""
    signal_freqs_ghz = GAIN_SIGNAL_FREQS_GHZ.copy()
    if enforce_pump_guard and not pump_is_stopband_safe(
        analysis_obj, float(pump_ghz)
    ):
        raise ValueError(
            f"Pump {pump_ghz:.4f} GHz is inside the configured "
            f"{PUMP_STOPBAND_GUARD_GHZ:.2f} GHz stopband guard."
        )

    valid, idler_freqs_ghz = valid_4wm_signal_mask(
        analysis_obj, pump_ghz, signal_freqs_ghz
    )
    raw_gain_db = np.full(signal_freqs_ghz.shape, np.nan)
    converged = np.zeros(signal_freqs_ghz.shape, dtype=bool)
    residuals = np.full(signal_freqs_ghz.shape, np.nan)
    passes_used = np.zeros(signal_freqs_ghz.shape, dtype=int)

    if np.any(valid):
        kwargs = dict(
            signal_freqs=signal_freqs_ghz[valid],
            Is0=float(Is0_value),
            pump=float(pump_ghz),
            model=GAIN_MODEL,
            modes=modes,
            thin=GAIN_SPATIAL_THIN,
            bloch_config=BLOCH_CONFIG,
        )
        if Ip0_value is not None:
            kwargs["Ip0"] = float(Ip0_value)
        result = analysis_obj.gain(**kwargs)
        result_gain = np.asarray(result.gain_db, dtype=float)
        result_converged = np.asarray(result.converged, dtype=bool)
        result_residuals = np.asarray(
            result.diagnostics.get(
                "residuals", np.full(len(result_gain), np.nan)
            ),
            dtype=float,
        )
        result_passes = np.asarray(
            result.diagnostics.get(
                "passes_used", np.zeros(len(result_gain))
            ),
            dtype=int,
        )
        raw_gain_db[valid] = result_gain
        converged[valid] = result_converged
        residuals[valid] = result_residuals
        passes_used[valid] = result_passes

    gain_db = raw_gain_db.copy()
    gain_db[~converged] = np.nan
    return {
        "signal_freqs": signal_freqs_ghz,
        "idler_freqs": idler_freqs_ghz,
        "valid": valid,
        "gain_db_raw": raw_gain_db,
        "gain_db": gain_db,
        "converged": converged,
        "residuals": residuals,
        "passes_used": passes_used,
    }


def plot_gain_curve(ax, curve, label, **plot_kwargs):
    """Solid converged curve plus explicit markers for finite failed points."""
    line, = ax.plot(
        curve["signal_freqs"], curve["gain_db"], label=label, **plot_kwargs
    )
    failed = (
        curve["valid"]
        & ~curve["converged"]
        & np.isfinite(curve["gain_db_raw"])
    )
    if np.any(failed):
        ax.scatter(
            curve["signal_freqs"][failed],
            curve["gain_db_raw"][failed],
            marker="x", s=20, alpha=0.75, color=line.get_color(),
            label="_nolegend_", zorder=4,
        )
    return line, failed


def shade_stopbands(ax, analysis_obj, pump_ghz=None):
    """Shade physical signal gaps and, optionally, their idler mirrors."""
    for band_index, (lower, upper) in enumerate(analysis_obj.linear.stopbands):
        ax.axvspan(
            lower, upper, color="tab:red", alpha=0.10,
            label="linear stopband" if band_index == 0 else "_nolegend_",
        )
        if pump_ghz is not None:
            ax.axvspan(
                2.0 * pump_ghz - upper, 2.0 * pump_ghz - lower,
                color="tab:orange", alpha=0.07,
                label=(
                    "idler maps into stopband"
                    if band_index == 0 else "_nolegend_"
                ),
            )


def print_gain_diagnostics(curve, label):
    """Print convergence totals and the two requested edge windows."""
    valid = curve["valid"]
    converged = curve["converged"]
    max_passes = int(np.max(curve["passes_used"][valid])) if np.any(valid) else 0
    print(
        f"{label}: {np.count_nonzero(converged)}/{np.count_nonzero(valid)} "
        f"valid points converged; maximum passes used = {max_passes}."
    )
    for lower, upper in ((1.0, 3.0), (16.0, 18.0)):
        in_window = (
            valid
            & (curve["signal_freqs"] >= lower)
            & (curve["signal_freqs"] <= upper)
        )
        print(
            f"  {lower:.0f}-{upper:.0f} GHz: "
            f"{np.count_nonzero(converged & in_window)}/"
            f"{np.count_nonzero(in_window)} converged"
        )


Ip0 = dBm_to_Ipeak(MAIN_PUMP_POWER_DBM, Z0_ref)
Is0 = dBm_to_Ipeak(MAIN_SIGNAL_POWER_DBM, Z0_ref)

print("\n========== PUMP / SIGNAL INPUT POWERS ==========")
print(f"Ppump input = {MAIN_PUMP_POWER_DBM:.2f} dBm")
print(f"Ip0 = {Ip0*1e3:.6f} mA")
print(f"Ip0/Istar = {Ip0/Istar:.6f}")
print(f"Check Ppump from Ip0 = {Ipeak_to_dBm(Ip0, Z0_ref):.2f} dBm")

print(f"\nPsignal input = {MAIN_SIGNAL_POWER_DBM:.2f} dBm")
print(f"Is0 = {Is0*1e6:.6f} uA")
print(f"Is0/Istar = {Is0/Istar:.6e}")
print(f"Check Psignal from Is0 = {Ipeak_to_dBm(Is0, Z0_ref):.2f} dBm")
print(f"Signal-to-pump current ratio Is0/Ip0 = {Is0/Ip0:.6e}")


# ============================================================
# 2. BUILD COSINE-MODULATED SUPERCELL
# ============================================================

n = np.arange(N_stubs_sc)

# Correct shifted cosine, no abs()
l_supercell = l_avg + 0.5 * l_mod * np.cos(2 * np.pi * n / N_stubs_sc)

# Total effective cell capacitance:
# C_cell = base capacitance + two stubs
#        = c' p + 2 c' l(x)
#
# For LCLfBaseCell, this is a lumped effective C per cell.
C_supercell = cprime * (p_cell + 2 * l_supercell)

# Average values
C_avg_cell = np.mean(C_supercell)
Lprime_avg = L_cell / p_cell
Cprime_avg = C_avg_cell / p_cell
v_avg = 1 / np.sqrt(Lprime_avg * Cprime_avg)
f_stop_est = v_avg / (2 * P_mod)
Z_avg = np.sqrt(L_cell / C_avg_cell)

print("\n========== GEOMETRY / AVERAGE LINE ==========")
print(f"Unit-cell pitch p_cell = {p_cell*1e6:.4f} um")
print(f"Cells per supercell = {N_stubs_sc}")
print(f"Modulation period P_mod = {P_mod*1e6:.4f} um")
print(f"Number of supercells = {N_supercells}")
print(f"Total cells = {N_stubs_sc * N_supercells}")
print(f"Total length = {device_length_mm:.4f} mm")

print("\n========== CELL PARAMETERS ==========")
print(f"L_cell = {L_cell*1e12:.4f} pH")
print(f"L' avg = {Lprime_avg*1e6:.4f} uH/m")
print(f"c' one-stub/strip = {cprime*1e9:.4f} nF/m")
print(f"C_avg_cell = {C_avg_cell*1e15:.4f} fF")
print(f"C' loaded avg = {Cprime_avg*1e9:.4f} nF/m")
print(f"Z_avg = {Z_avg:.4f} ohm")
print(f"v_avg = {v_avg:.4e} m/s = {v_avg/3e8:.5f} c")
print(f"Bragg estimate f_stop = {f_stop_est/1e9:.6f} GHz")

print("\n========== STUB LENGTHS ==========")
print(f"l_min = {np.min(l_supercell)*1e6:.4f} um")
print(f"l_avg = {np.mean(l_supercell)*1e6:.4f} um")
print(f"l_max = {np.max(l_supercell)*1e6:.4f} um")


# ============================================================
# 3. BUILD TWPASOLVER MODEL
# ============================================================

# Each cell is a series L plus shunt effective C with optional Lf.
# This is the LCLfBaseCell model from your existing notebook.
cells = [
    LCLfBaseCell(
        L=float(L_cell),
        C=float(C_supercell[i]),
        Lf=float(Lf_stub),
        delta=float(DIELECTRIC_LOSS_TANGENT),
        qp_loss=float(QP_LOSS_PER_HZ),
        N=1,
        Z0_ref=Z0_ref,
    )
    for i in range(N_stubs_sc)
]

twpa = TWPA(
    cells=cells,
    N=N_supercells,
    Istar=Istar,
    Idc=Idc,
    Ip0=Ip0,
    Z0_ref=Z0_ref,
)

print("\n========== TWPASOLVER MODEL ==========")
print(twpa.model_dump())


# ============================================================
# 4. LINEAR RESPONSE AND BANDGAP
# ============================================================

net = twpa.get_network(freqs_hz)

plt.figure(figsize=(9, 5))
net.s21.plot_s_db(label="Cosine modulated TWPA", lw=2)
plt.axvline(f_stop_est / 1e9, ls="--", label="v/(2Pmod) estimate")
plt.xlabel("Frequency (GHz)")
plt.ylabel("|S21| (dB)")
plt.title(
    "Linear response / stopband\n"
    f"N = {N_supercells}, L = {device_length_mm:.2f} mm, "
    f"fstop,Bragg ≈ {f_stop_est/1e9:.3f} GHz"
)
plt.grid(True)
plt.legend()
plt.xlim(0, 18e9)
plt.ylim(-40, 5)
plt.show()


def get_s21_db(network):
    """
    Try to access S21 dB numerically.
    Attribute names may differ slightly depending on twpasolver/scikit-rf version.
    """
    s21 = network.s21

    for attr in ["s_db", "s_db_unwrap"]:
        if hasattr(s21, attr):
            return np.array(getattr(s21, attr)).flatten()

    # fallback: scikit-rf Network often has .s complex array
    if hasattr(s21, "s"):
        return 20 * np.log10(np.abs(np.array(s21.s).flatten()))

    raise AttributeError("Could not find S21 dB data on network.s21")


s21_db = get_s21_db(net)

# Save the data used by the direct network S21 plot.
combined = np.column_stack((freqs_hz / 1e9, s21_db))
linear_s21_filename = (
    f"linear_network_s21_N{N_supercells}_"
    f"length{device_length_mm:.3f}mm.csv"
)
# np.savetxt(f"C:\\Users\\Lamon\\TWPA_solver\\{linear_s21_filename}", combined, delimiter=",", header="frequency_ghz,s21_db", comments="", fmt="%g")
# np.savetxt(f"G:\\My Drive\\twpasolver_data\\{linear_s21_filename}", combined, delimiter=",", header="frequency_ghz,s21_db", comments="", fmt="%g")

# Find deepest stopband near the Bragg estimate
search_lo = 0.5 * f_stop_est
search_hi = 1.5 * f_stop_est
idx = np.where((freqs_hz >= search_lo) & (freqs_hz <= search_hi))[0]

idx_stop = idx[np.argmin(s21_db[idx])]
f_stop_s21 = freqs_hz[idx_stop]

print("\n========== BANDGAP / STOPBAND ==========")
print(f"Bragg estimate f_stop = {f_stop_est/1e9:.6f} GHz")
print(f"S21 minimum near estimate = {f_stop_s21/1e9:.6f} GHz")
print(f"S21 at stopband = {s21_db[idx_stop]:.4f} dB")


# ============================================================
# PLOT TITLE / CONTEXT STRINGS
# ============================================================

context_line_1 = (
    f"N = {N_supercells}, L = {device_length_mm:.2f} mm, "
    f"Ppump = {MAIN_PUMP_POWER_DBM:.1f} dBm, Psignal = {MAIN_SIGNAL_POWER_DBM:.1f} dBm"
)

context_line_2 = (
    f"Ip0/I* = {Ip0/Istar:.3f}, Is0/Ip0 = {Is0/Ip0:.2e}, "
    f"fstop ≈ {f_stop_s21/1e9:.3f} GHz"
)

plot_context = context_line_1 + "\n" + context_line_2


# ============================================================
# 5. TWPAnalysis BASE DATA
# ============================================================

# TWPAnalysis expects frequencies in GHz by default when using f_arange.
# Use a range that includes signal, pump, idler.
#
# For basic 4WM with fp ~ 10.6 GHz and fs = 4–8 GHz,
# idler = 2fp - fs ~ 13.2–17.2 GHz, so 30 GHz is fine.
#
# If you later include higher products like 3fp or 4fp-fs,
# increase this to ~50 GHz.
analysis = TWPAnalysis(
    twpa=twpa,
    freqs=ANALYSIS_FREQUENCY_RANGE_GHZ,
    unit="GHz",
)

linear = analysis.linear  # optional: explicitly trigger calculation

# The two TWPAnalysis linear-response panels share this frequency grid.
combined = np.column_stack((linear.freqs, linear.S21_db, linear.k_star))
linear_response_filename = (
    f"linear_analysis_response_N{N_supercells}_"
    f"length{device_length_mm:.3f}mm.csv"
)
# np.savetxt(f"C:\\Users\\Lamon\\TWPA_solver\\{linear_response_filename}", combined, delimiter=",", header="frequency_ghz,s21_db,k_star_rad", comments="", fmt="%g")
# np.savetxt(f"G:\\My Drive\\twpasolver_data\\{linear_response_filename}", combined, delimiter=",", header="frequency_ghz,s21_db,k_star_rad", comments="", fmt="%g")

response_pump_marker = 0.85 * f_stop_est / 1e9

ax = analysis.plot_response(pump_freq=response_pump_marker)
plt.suptitle(
    "TWPAnalysis linear response\n"
    f"Pump marker = {response_pump_marker:.3f} GHz, "
    f"Ppump = {MAIN_PUMP_POWER_DBM:.1f} dBm, N = {N_supercells}, L = {device_length_mm:.2f} mm",
    y=1.02
)
plt.show()


# ============================================================
# 6. BASIC 4WM MODE ARRAY
# ============================================================
#
# Latest twpasolver API:
#   ModeArrayFactory.create_basic(..., three_wave=False)
#
# Physical interpretation:
#   p = pump
#   s = signal
#   i = idler
#
# Relation:
#   f_i = 2 f_p - f_s
#
# This is the clean/basic 4WM p-s-i model.
# It does NOT include p2/pp, p3, 2p+s, etc.

mode_array_config_name = "basic_4wm"

mode_array_4wm = ModeSet.basic(three_wave=False)   # False selects 4WM: i = p + p - s



print("\n========== BASIC 4WM MODE ARRAY ==========")
print(f"mode_array_config_name = {mode_array_config_name}")
print(f"gain_model = {GAIN_MODEL}")

# mode_array_4wm.print_symbolic_expressions()

print("\n========== BASIC 4WM RWA TERMS ==========")
terms_4wm = mode_array_4wm.rwa_terms(power=3)

print(f"Number of 4WM RWA terms found: {len(terms_4wm)}")
for term in terms_4wm:
    print(term)


# ============================================================
# RESIDUAL UNWRAPPED S21 PHASE, CALCULATED DIRECTLY FROM S21
# ============================================================

# Complex S21 data from the TWPAnalysis circuit response.
freqs_ghz = np.asarray(analysis.freqs) #already in GHz bc we used unit="GHz" in TWPAnalysis
s21 = np.asarray(analysis.linear.S21).squeeze()

# 1. Unwrap the total transmission phase.
phase_unwrapped = np.unwrap(np.angle(s21))

# 2. Locate the lower edge of the stopband using the steepest
#    decrease in |S21|. Multiplying by log10 instead of ln does not
#    affect the detected index.
s21_db = 20 * np.log10(
    np.clip(np.abs(s21), 1e-300, None)
)

stopband_start_idx = np.argmin(np.diff(s21_db))

# Use the same low-frequency fitting region as TWPAnalysis:
# frequencies from the beginning through approximately one tenth
# of the stopband-start index.
fit_end_idx = max(int(stopband_start_idx / 10), 3)

fit_freqs = freqs_ghz[:fit_end_idx]
fit_phase = phase_unwrapped[:fit_end_idx]

# 3. Fit:
#
# phase(f) = a*f^2 + b*f + c
#
# The quadratic fit makes the extracted linear slope less sensitive
# to weak low-frequency curvature.
phase_coefficients = np.polyfit(
    fit_freqs,
    fit_phase,
    deg=2,
)

a, b, c = phase_coefficients

# 4. Construct only the fitted linear baseline b*f + c.
phase_linear = b * freqs_ghz + c

# 5. Subtract the linear baseline from the unwrapped phase.
phase_residual = phase_unwrapped - phase_linear

# ============================================================
# PLOT
# ============================================================

plt.figure(figsize=(9, 5))

plt.plot(
    freqs_ghz,
    phase_residual,
    lw=2,
    label=r"$\mathrm{unwrap}(\angle S_{21})-(bf+c)$",
)

plt.axhline(
    0,
    color="black",
    linestyle="--",
    linewidth=1,
)

# Show the frequency interval used to obtain the linear fit.
plt.axvspan(
    fit_freqs[0],
    fit_freqs[-1],
    color="tab:blue",
    alpha=0.12,
    label="Low-frequency fit region",
)

# Add the detected stopband to titles when it is available.
if "f_stop_s21" in globals():
    plt.axvline(
        f_stop_s21 / 1e9,
        color="tab:red",
        linestyle=":",
        linewidth=2,
        label="Stopband",
    )

plt.xlabel("Frequency [GHz]")
plt.ylabel("Residual unwrapped phase [rad]")
plt.title("Unwrapped transmission phase with linear trend removed")
plt.xlim(10, 18)   
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

print("Phase fit coefficients:")
print(f"  quadratic a = {a:.6e} rad/GHz²")
print(f"  linear    b = {b:.6e} rad/GHz")
print(f"  intercept c = {c:.6e} rad")
print(
    f"Fit interval: {fit_freqs[0]:.4f} to "
    f"{fit_freqs[-1]:.4f} GHz"
)

# ============================================================
# RESIDUAL UNWRAPPED S21 PHASE IN DEGREES
# ============================================================

freqs_ghz = np.asarray(analysis.freqs)
s21 = np.asarray(analysis.linear.S21).squeeze()

# 1. Unwrap in radians, then convert to degrees.
phase_unwrapped_deg = np.rad2deg(
    np.unwrap(np.angle(s21))
)

# 2. Locate the lower stopband edge.
s21_db = 20 * np.log10(
    np.clip(np.abs(s21), 1e-300, None)
)

stopband_start_idx = np.argmin(np.diff(s21_db))

# Same low-frequency fitting region used previously.
fit_end_idx = max(int(stopband_start_idx / 10), 3)

fit_freqs = freqs_ghz[:fit_end_idx]
fit_phase_deg = phase_unwrapped_deg[:fit_end_idx]

# 3. Fit:
#
# phase(f) = a*f^2 + b*f + c
#
# The coefficients now have units based on degrees.
phase_coefficients_deg = np.polyfit(
    fit_freqs,
    fit_phase_deg,
    deg=2,
)

a_deg, b_deg, c_deg = phase_coefficients_deg

# 4. Linear baseline in degrees.
phase_linear_deg = b_deg * freqs_ghz + c_deg

# 5. Residual unwrapped phase in degrees.
phase_residual_deg = (
    phase_unwrapped_deg
    - phase_linear_deg
)

# Save the radian and degree phase plots together because they share a grid.
combined = np.column_stack((
    freqs_ghz, s21_db, phase_unwrapped, phase_linear, phase_residual,
    phase_unwrapped_deg, phase_linear_deg, phase_residual_deg,
))
linear_phase_filename = (
    f"linear_residual_phase_N{N_supercells}_"
    f"length{device_length_mm:.3f}mm.csv"
)
linear_phase_header = (
    "frequency_ghz,s21_db,phase_unwrapped_rad,phase_linear_rad,"
    "phase_residual_rad,phase_unwrapped_deg,phase_linear_deg,"
    "phase_residual_deg"
)
# np.savetxt(f"C:\\Users\\Lamon\\TWPA_solver\\{linear_phase_filename}", combined, delimiter=",", header=linear_phase_header, comments="", fmt="%g")
# np.savetxt(f"G:\\My Drive\\twpasolver_data\\{linear_phase_filename}", combined, delimiter=",", header=linear_phase_header, comments="", fmt="%g")

# ============================================================
# PLOT
# ============================================================

plt.figure(figsize=(9, 5))

plt.plot(
    freqs_ghz,
    phase_residual_deg,
    lw=2,
    label=r"$\mathrm{unwrap}(\angle S_{21})-(bf+c)$",
)

plt.axhline(
    0,
    color="black",
    linestyle="--",
    linewidth=1,
)

plt.axvspan(
    fit_freqs[0],
    fit_freqs[-1],
    color="tab:blue",
    alpha=0.12,
    label="Low-frequency fit region",
)

if "f_stop_s21" in globals():
    plt.axvline(
        f_stop_s21 / 1e9,
        color="tab:red",
        linestyle=":",
        linewidth=2,
        label="Stopband",
    )

plt.xlabel("Frequency [GHz]")
plt.ylabel(r"Residual unwrapped phase [$^\circ$]")
plt.title(
    "Unwrapped transmission phase with linear trend removed"
)
plt.xlim(10, 18)
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

print("Phase fit coefficients:")
print(
    f"  quadratic a = {a_deg:.6e} degree/GHz²"
)
print(
    f"  linear    b = {b_deg:.6e} degree/GHz"
)
print(
    f"  intercept c = {c_deg:.6e} degree"
)
print(
    f"Fit interval: {fit_freqs[0]:.4f} to "
    f"{fit_freqs[-1]:.4f} GHz"
)
# ============================================================
# 7. PHASE MATCHING: PUMP SWEEP, SIGNAL WINDOW 4–8 GHz
# ============================================================

signal_freqs_pm = inclusive_frequency_grid(
    PHASE_MATCH_SIGNAL_MIN_GHZ,
    PHASE_MATCH_SIGNAL_MAX_GHZ,
    GAIN_FREQUENCY_STEP_GHZ,
)
pump_freqs_pm = inclusive_frequency_grid(
    PHASE_MATCH_PUMP_MIN_GHZ,
    PHASE_MATCH_PUMP_MAX_GHZ,
    GAIN_FREQUENCY_STEP_GHZ,
)

pm = analysis.phase_matching(
    mixing="4WM",
    process="PA",
    pump_mode="p",
    signal_mode="s",
    idler_mode="i",
    modes="basic_4wm",
    pump_Ip0=Ip0,
    signal_freqs=signal_freqs_pm,
    pump_freqs=pump_freqs_pm,
    relative=True,
)

# Save the phase-matching heatmap in long form (one row per grid point).
pm_pump_mesh, pm_signal_mesh = np.meshgrid(pm.pump_freqs, pm.signal_freqs)
pm_delta = np.asarray(pm.delta, dtype=float)
with np.errstate(divide="ignore", invalid="ignore"):
    pm_log_abs_delta = np.log(np.abs(pm_delta))
combined = np.column_stack((
    pm_pump_mesh.ravel(), pm_signal_mesh.ravel(),
    pm_delta.ravel(), pm_log_abs_delta.ravel(),
))
phase_heatmap_filename = (
    f"phase_matching_heatmap_Ppump{MAIN_PUMP_POWER_DBM:.1f}dBm_"
    f"Ip0_Istar{Ip0/Istar:.6f}_N{N_supercells}.csv"
)
phase_heatmap_header = (
    "pump_frequency_ghz,signal_frequency_ghz,"
    "delta_k_over_ks,log_abs_delta_k_over_ks"
)
# np.savetxt(f"C:\\Users\\Lamon\\TWPA_solver\\{phase_heatmap_filename}", combined, delimiter=",", header=phase_heatmap_header, comments="", fmt="%g")
# np.savetxt(f"G:\\My Drive\\twpasolver_data\\{phase_heatmap_filename}", combined, delimiter=",", header=phase_heatmap_header, comments="", fmt="%g")

ax = pm.plot()
ax.collections[0].colorbar.set_label(
    r"$\ln|\Delta k/k_s|$"
)
plt.title(
    r"Basic 4WM normalized phase matching: $\ln|\Delta k/k_s|$"
    "\n"
    f"Ppump = {MAIN_PUMP_POWER_DBM:.1f} dBm, "
    f"N = {N_supercells}, L = {device_length_mm:.2f} mm, "
    f"Ip0/I* = {Ip0/Istar:.3f}, fstop ≈ {f_stop_s21/1e9:.3f} GHz"
)
plt.xlabel("Pump frequency [GHz]")
plt.ylabel("Signal frequency [GHz]")
plt.xticks(np.arange(pump_freqs_pm[0], pump_freqs_pm[-1] + 0.5, 0.5), rotation=90)
plt.yticks(np.arange(signal_freqs_pm[0], signal_freqs_pm[-1] + 0.5, 0.5))

plt.show()

print("\n========== PHASE MATCHING ==========")
print(f"Signal window = {signal_freqs_pm[0]:.3f} to {signal_freqs_pm[-1]:.3f} GHz")
print(f"Pump sweep = {pump_freqs_pm[0]:.3f} to {pump_freqs_pm[-1]:.3f} GHz")
# print("Phase matching keys:", list(pm.keys()))


# ============================================================
# 8. 4WM GAIN USING CMES: SELECTED PUMP FREQUENCIES
# ============================================================

# The old 14.50/14.56 GHz points are deliberately retained as candidates:
# the guard reports and skips them instead of spending minutes on a pump
# whose driven Bloch basis is unreliable at the stopband edge.
pump_list = np.array([
    pump for pump in SELECTED_PUMP_CANDIDATES_GHZ
    if pump_is_stopband_safe(analysis, float(pump))
])
skipped_pumps = np.setdiff1d(SELECTED_PUMP_CANDIDATES_GHZ, pump_list)
if len(skipped_pumps):
    print(
        "Skipped pump frequencies inside the stopband guard: "
        + ", ".join(f"{pump:.3f} GHz" for pump in skipped_pumps)
    )

fig, ax = plt.subplots(figsize=(10, 5.5))
gain_results = {}
has_failed_points = False
for pump_freq in pump_list:
    curve = run_gain_curve(
        analysis,
        Is0,
        float(pump_freq),
        modes="basic_4wm",
    )
    gain_results[float(pump_freq)] = curve
    _, failed = plot_gain_curve(
        ax, curve, label=f"{pump_freq:.2f} GHz", linewidth=1.6
    )
    has_failed_points |= bool(np.any(failed))
    print_gain_diagnostics(curve, f"Pump {pump_freq:.3f} GHz")

shade_stopbands(ax, analysis)
if has_failed_points:
    ax.scatter(
        [], [], marker="x", color="black", s=25,
        label="finite but not converged (diagnostic only)",
    )
ax.set_xlabel("Signal frequency [GHz]")
ax.set_ylabel("Signal gain S21 [dB]")
ax.set_title(
    "Basic 4WM gain for selected pump frequencies\n"
    f"Ppump = {MAIN_PUMP_POWER_DBM:.1f} dBm, Psignal = {MAIN_SIGNAL_POWER_DBM:.1f} dBm, "
    f"model = {GAIN_MODEL}"
    "\n"
    f"N = {N_supercells}, L = {device_length_mm:.2f} mm, "
    f"Ip0/I* = {Ip0/Istar:.3f}, fstop ≈ {f_stop_s21/1e9:.3f} GHz"
)
ax.set_xlim(GAIN_SIGNAL_MIN_GHZ, GAIN_SIGNAL_MAX_GHZ)
ax.grid(True, alpha=0.3)
ax.legend(title="Pump")
fig.tight_layout()
plt.show()


# ============================================================
# 9. PRINT SUMMARY OF SELECTED PUMP RESULTS
# ============================================================

print("\n========== BASIC 4WM GAIN SUMMARY ==========")
print(f"Gain model = {GAIN_MODEL}")
print(f"Ppump = {MAIN_PUMP_POWER_DBM:.2f} dBm | Ip0 = {Ip0*1e3:.6f} mA | Ip0/Istar = {Ip0/Istar:.6f}")
print(f"Psignal = {MAIN_SIGNAL_POWER_DBM:.2f} dBm | Is0 = {Is0*1e6:.6f} uA | Is0/Ip0 = {Is0/Ip0:.6e}")
print(f"Estimated stopband = {f_stop_s21/1e9:.6f} GHz")

for pump_freq, curve in gain_results.items():
    gain_db = curve["gain_db"]
    signal_freqs = curve["signal_freqs"]    
    finite = np.isfinite(gain_db)

    if not finite.any():
        print(
            f"Pump = {pump_freq:.3f} GHz | "
            "no valid converged gain points"
        )
        continue
    # Combine into a 2D array with 2 columns and save
    combined = np.column_stack((signal_freqs, gain_db))
    # np.savetxt(f"C:\\Users\\Lamon\\TWPA_solver\\gain_pumpfreq{pump_freq:.3f}_Ip0_Istar{Ip0/Istar:.6f}.csv", combined, delimiter=",", header="signal_freq,gain_db", comments="", fmt="%g")
    # np.savetxt(f"G:\\My Drive\\twpasolver_data\\gain_pumpfreq{pump_freq:.3f}_Ip0_Istar{Ip0/Istar:.6f}.csv", combined, delimiter=",", header="signal_freq,gain_db", comments="", fmt="%g")
    max_idx = int(np.nanargmax(gain_db))

    print(
        f"Pump = {pump_freq:6.3f} GHz | "
        f"Max gain = {gain_db[max_idx]:8.3f} dB at "
        f"signal = {curve['signal_freqs'][max_idx]:6.3f} GHz | "
        f"Mean gain = {np.nanmean(gain_db):8.3f} dB"
    )

# ============================================================
# 10. CURRENT MODE ANALYSIS FOR ONE PUMP
# ============================================================

print("\n=========== CURRENT MODE ANALYSIS =============")

pump_freq_current_plot = CURRENT_PROFILE_PUMP_GHZ
signal_freq_current_plot = CURRENT_PROFILE_SIGNAL_GHZ

gain_res_current = analysis.gain(
    signal_freqs=[signal_freq_current_plot],
    Is0=Is0,
    pump=pump_freq_current_plot,
    model=GAIN_MODEL,
    modes="basic_4wm",
    thin=GAIN_SPATIAL_THIN,
    bloch_config=BLOCH_CONFIG,
)

ax = gain_res_current.plot_currents()
ax.set_title(
    "Basic 4WM forward-mode currents\n"
    f"fp = {pump_freq_current_plot:.3f} GHz, "
    f"fs = {signal_freq_current_plot:.3f} GHz, "
    f"Ppump = {MAIN_PUMP_POWER_DBM:.1f} dBm"
)
plt.show()

if gain_res_current.model == "bloch":
    ax = gain_res_current.plot_currents(backward=True)
    ax.set_title(
        "Basic 4WM backward-mode currents\n"
        f"fp = {pump_freq_current_plot:.3f} GHz, "
        f"fs = {signal_freq_current_plot:.3f} GHz"
    )
    plt.show()
```

```python
# ============================================================
# HEATMAP: MAXIMUM GAIN VS PUMP FREQUENCY AND PUMP POWER
# ============================================================
#
# Run the main setup block before running this cell.
#
# Heatmap axes:
#   x = pump frequency [GHz]
#   y = pump power [dBm]
#   color = maximum converged gain inside the configured 4-8 GHz
#           heatmap evaluation band [dB]
#
# Every solve still uses the shared full gain grid. Only points inside
# HEATMAP_EVALUATION_MIN_GHZ to HEATMAP_EVALUATION_MAX_GHZ are used
# to score the heatmap maximum and convergence fraction.

# ------------------------------------------------------------
# Settings are defined in the top workflow-settings section.
# ------------------------------------------------------------

pump_powers_dbm = HEATMAP_PUMP_POWERS_DBM

# Use the lower edge extracted by TWPAnalysis rather than the
# frequency of the deepest point in the stopband.
stopbands = analysis.linear.stopbands

if len(stopbands) == 0:
    raise RuntimeError("No stopband was detected in the analysis frequency grid.")

stopband_lower_ghz, stopband_upper_ghz = map(float, stopbands[0])

print(
    f"First stopband: "
    f"{stopband_lower_ghz:.4f}–{stopband_upper_ghz:.4f} GHz"
)
HEATMAP_PUMP_STEP_GHZ = 0.1
pump_frequency_candidates_ghz = inclusive_frequency_grid(
    HEATMAP_PUMP_MIN_GHZ,
    HEATMAP_PUMP_MAX_GHZ,
    HEATMAP_PUMP_STEP_GHZ,
)
pump_frequencies_ghz = np.array([
    pump for pump in pump_frequency_candidates_ghz
    if pump_is_stopband_safe(analysis, float(pump))
])
signal_frequencies_sweep = GAIN_SIGNAL_FREQS_GHZ
heatmap_evaluation_mask = (
    (signal_frequencies_sweep >= HEATMAP_EVALUATION_MIN_GHZ)
    & (signal_frequencies_sweep <= HEATMAP_EVALUATION_MAX_GHZ)
)
if not np.any(heatmap_evaluation_mask):
    raise RuntimeError("The heatmap evaluation band contains no gain-grid points.")

# Use the signal power defined at the top of the notebook.
Is0_sweep = dBm_to_Ipeak(
    HEATMAP_SIGNAL_POWER_DBM,
    Z0_ref,
)

print("========== MAX-GAIN HEATMAP SETTINGS ==========")
print(
    f"Gain model: {GAIN_MODEL}"
)
print(
    f"Signal power: {HEATMAP_SIGNAL_POWER_DBM:.2f} dBm"
)
print(
    f"Lower stopband edge: "
    f"{stopband_lower_ghz:.4f} GHz"
)
print(
    f"Full solved signal range: "
    f"{signal_frequencies_sweep[0]:.3f} to "
    f"{signal_frequencies_sweep[-1]:.3f} GHz"
)
print(
    f"Heatmap maximum evaluated only from "
    f"{HEATMAP_EVALUATION_MIN_GHZ:.3f} to "
    f"{HEATMAP_EVALUATION_MAX_GHZ:.3f} GHz"
)
print(
    f"Pump-frequency range: "
    f"{pump_frequencies_ghz[0]:.3f} to "
    f"{pump_frequencies_ghz[-1]:.3f} GHz"
)
print(
    f"Pump-power range: "
    f"{pump_powers_dbm[0]:.1f} to "
    f"{pump_powers_dbm[-1]:.1f} dBm"
)

# ------------------------------------------------------------
# Result arrays
# ------------------------------------------------------------

# Shape: [pump_power_index, pump_frequency_index]
max_gain_db_map = np.full(
    (
        len(pump_powers_dbm),
        len(pump_frequencies_ghz),
    ),
    np.nan,
)

# Store the signal frequency where maximum gain occurs.
max_gain_signal_ghz_map = np.full_like(
    max_gain_db_map,
    np.nan,
)

# Fraction of valid evaluation-band frequencies that converged.
converged_fraction_map = np.full_like(
    max_gain_db_map,
    np.nan,
)

# ------------------------------------------------------------
# Sweep pump power
# ------------------------------------------------------------

for power_index, pump_power_dbm in enumerate(
    pump_powers_dbm
):
    Ip0_this = dBm_to_Ipeak(
        pump_power_dbm,
        Z0_ref,
    )

    print(
        f"\nPump power {pump_power_dbm:.1f} dBm "
        f"({power_index + 1}/{len(pump_powers_dbm)}), "
        f"Ip0/Istar = {Ip0_this / Istar:.4f}"
    )
    if Ip0_this / Istar > HEATMAP_MAX_PUMP_CURRENT_RATIO:
        print(
            f"Point skipped: Ip0/Istar > "
            f"{HEATMAP_MAX_PUMP_CURRENT_RATIO:.2f}"
        )
        continue

    # Reuse the main analysis object.  Ip0 is an explicit gain() input, so
    # rebuilding TWPA, its network, and its linear tables for every power
    # is unnecessary.

    # --------------------------------------------------------
    # Sweep pump frequency at this pump power
    # --------------------------------------------------------

    for pump_index, pump_frequency in enumerate(
        pump_frequencies_ghz
    ):
        curve = run_gain_curve(
            analysis,
            Is0_sweep,
            float(pump_frequency),
            Ip0_value=Ip0_this,
            modes="basic_4wm",
        )

        gain_db = curve["gain_db"]
        evaluation_valid = curve["valid"] & heatmap_evaluation_mask
        evaluation_converged = evaluation_valid & curve["converged"]
        n_valid = np.count_nonzero(evaluation_valid)
        converged_fraction_map[power_index, pump_index] = (
            np.count_nonzero(evaluation_converged) / n_valid
            if n_valid else np.nan
        )

        finite_gain = evaluation_converged & np.isfinite(gain_db)

        if not np.any(finite_gain):
            continue

        finite_indices = np.where(finite_gain)[0]

        local_max_index = finite_indices[
            np.argmax(gain_db[finite_gain])
        ]

        max_gain_db_map[
            power_index,
            pump_index,
        ] = gain_db[local_max_index]

        max_gain_signal_ghz_map[
            power_index,
            pump_index,
        ] = curve["signal_freqs"][local_max_index]

# Both heatmaps share the same pump-frequency/pump-power grid.
heatmap_pump_mesh, heatmap_power_mesh = np.meshgrid(
    pump_frequencies_ghz, pump_powers_dbm
)
combined = np.column_stack((
    heatmap_pump_mesh.ravel(), heatmap_power_mesh.ravel(),
    max_gain_db_map.ravel(), max_gain_signal_ghz_map.ravel(),
    converged_fraction_map.ravel(),
))
gain_heatmap_filename = (
    f"gain_pump_frequency_power_heatmap_"
    f"Psignal{HEATMAP_SIGNAL_POWER_DBM:.1f}dBm_"
    f"Istar{Istar * 1e3:.3f}mA_N{N_supercells}.csv"
)
gain_heatmap_header = (
    "pump_frequency_ghz,pump_power_dbm,max_gain_db,"
    "max_gain_signal_frequency_ghz,converged_fraction"
)
# np.savetxt(f"C:\\Users\\Lamon\\TWPA_solver\\{gain_heatmap_filename}", combined, delimiter=",", header=gain_heatmap_header, comments="", fmt="%g")
# np.savetxt(f"G:\\My Drive\\twpasolver_data\\{gain_heatmap_filename}", combined, delimiter=",", header=gain_heatmap_header, comments="", fmt="%g")

# ============================================================
# PLOT HEATMAP
# ============================================================

fig, ax = plt.subplots(
    figsize=(10, 6),
)

mesh = ax.pcolormesh(
    pump_frequencies_ghz,
    pump_powers_dbm,
    max_gain_db_map,
    shading="auto",
    cmap="viridis",
)

colorbar = fig.colorbar(
    mesh,
    ax=ax,
)

colorbar.set_label(
    f"Maximum converged gain in {HEATMAP_EVALUATION_MIN_GHZ:g}-"
    f"{HEATMAP_EVALUATION_MAX_GHZ:g} GHz [dB]"
)

ax.set_xlabel(
    "Pump frequency [GHz]"
)
ax.set_ylabel(
    "Pump power [dBm]"
)
ax.set_title(
    f"Basic 4WM maximum converged gain in {HEATMAP_EVALUATION_MIN_GHZ:g}-"
    f"{HEATMAP_EVALUATION_MAX_GHZ:g} GHz\n"
    f"Psignal = {HEATMAP_SIGNAL_POWER_DBM:.1f} dBm, "
    f"model = {GAIN_MODEL}, "
    f"fstop,lower = {stopband_lower_ghz:.3f} GHz"
)

# Mark the global maximum.
if np.any(np.isfinite(max_gain_db_map)):
    flat_max_index = np.nanargmax(
        max_gain_db_map
    )

    best_power_index, best_pump_index = (
        np.unravel_index(
            flat_max_index,
            max_gain_db_map.shape,
        )
    )

    best_pump_power = pump_powers_dbm[
        best_power_index
    ]
    best_pump_frequency = pump_frequencies_ghz[
        best_pump_index
    ]
    best_gain = max_gain_db_map[
        best_power_index,
        best_pump_index,
    ]
    best_signal_frequency = (
        max_gain_signal_ghz_map[
            best_power_index,
            best_pump_index,
        ]
    )

    ax.scatter(
        best_pump_frequency,
        best_pump_power,
        marker="*",
        s=180,
        color="red",
        edgecolor="white",
        linewidth=1,
        label=(
            f"Maximum in {HEATMAP_EVALUATION_MIN_GHZ:g}-"
            f"{HEATMAP_EVALUATION_MAX_GHZ:g} GHz: {best_gain:.2f} dB\n"
            f"fp={best_pump_frequency:.2f} GHz, "
            f"fs={best_signal_frequency:.2f} GHz"
        ),
    )

    ax.legend(
        loc="best",
    )

    print(
        f"\n========== GLOBAL MAXIMUM IN "
        f"{HEATMAP_EVALUATION_MIN_GHZ:g}-"
        f"{HEATMAP_EVALUATION_MAX_GHZ:g} GHz =========="
    )
    print(
        f"Pump power: {best_pump_power:.2f} dBm"
    )
    print(
        f"Pump frequency: "
        f"{best_pump_frequency:.4f} GHz"
    )
    print(
        f"Signal frequency: "
        f"{best_signal_frequency:.4f} GHz"
    )
    print(
        f"Maximum gain: {best_gain:.4f} dB"
    )

plt.tight_layout()
plt.show()

# A second map prevents a high-gain but poorly converged operating point
# from looking equivalent to a numerically clean one.
fig_conv, ax_conv = plt.subplots(figsize=(10, 5.5))
mesh_conv = ax_conv.pcolormesh(
    pump_frequencies_ghz,
    pump_powers_dbm,
    100.0 * converged_fraction_map,
    shading="auto",
    cmap="magma",
    vmin=0.0,
    vmax=100.0,
)
fig_conv.colorbar(
    mesh_conv, ax=ax_conv,
    label=(
        f"Converged valid {HEATMAP_EVALUATION_MIN_GHZ:g}-"
        f"{HEATMAP_EVALUATION_MAX_GHZ:g} GHz points [%]"
    ),
)
ax_conv.set_xlabel("Pump frequency [GHz]")
ax_conv.set_ylabel("Pump power [dBm]")
ax_conv.set_title(
    f"Bloch-solver convergence in heatmap evaluation band: "
    f"{HEATMAP_EVALUATION_MIN_GHZ:g}-"
    f"{HEATMAP_EVALUATION_MAX_GHZ:g} GHz"
)
fig_conv.tight_layout()
plt.show()
```

```python
# ============================================================
# ISTAR SWEEP: BASIC 4WM GAIN VS SIGNAL FREQUENCY
# ============================================================
#
# This mirrors the pump-power sweep, but holds pump power fixed.
# Consistency with the original model is explicit in this block:
#   - N is passed directly as N_supercells.
#   - Istar values are entered in amperes and passed directly to TWPA.
#   - Ip0 uses the same dBm_to_Ipeak(..., Z0_ref) conversion as the
#     top setup and each point of the pump-power sweep.
# A new TWPA/TWPAnalysis is still required for every Istar because
# Istar changes the nonlinear coefficient used by the CME solver.

import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. USER-EDITABLE ISTAR-SWEEP SETTINGS
# ============================================================

# Enter Istar directly in amperes, matching Istar in the top cell.
# These exact values are passed to TWPA without rescaling or casting.
ISTAR_SWEEP_VALUES_A = (
    7e-3,
    7.5e-3,
    8e-3,
    8.5e-3,
    9e-3,
)

# Choose the one pump power used for every Istar point.
ISTAR_SWEEP_FIXED_PUMP_POWER_DBM = -15.0

# Inherit pump frequency and signal power from the pump-power sweep.
ISTAR_SWEEP_PUMP_GHZ = PUMP_POWER_SWEEP_PUMP_GHZ
ISTAR_SWEEP_SIGNAL_POWER_DBM = PUMP_POWER_SWEEP_SIGNAL_POWER_DBM

# ============================================================
# 2. VALIDATE INPUTS AND CREATE DISPLAY-ONLY VALUES
# ============================================================

# This copy is only for validation and plotting; it is not passed to TWPA.
istar_sweep_values_for_checks = np.asarray(ISTAR_SWEEP_VALUES_A, dtype=float)
if istar_sweep_values_for_checks.size == 0:
    raise ValueError("ISTAR_SWEEP_VALUES_A must contain at least one value.")
if (
    np.any(~np.isfinite(istar_sweep_values_for_checks))
    or np.any(istar_sweep_values_for_checks <= 0)
):
    raise ValueError("All Istar sweep values must be finite and greater than zero.")

# Convert to mA only for labels, axes, and printed output.
istar_sweep_values_ma = istar_sweep_values_for_checks * 1e3

# ============================================================
# 3. CALCULATE FIXED PUMP AND SIGNAL CURRENTS
# ============================================================

# This is the exact Ip0 conversion used in the top and pump-sweep cells.
istar_sweep_Ip0 = dBm_to_Ipeak(
    ISTAR_SWEEP_FIXED_PUMP_POWER_DBM, Z0_ref
)

# Use the same signal-current conversion as the other gain blocks.
istar_sweep_Is0 = dBm_to_Ipeak(
    ISTAR_SWEEP_SIGNAL_POWER_DBM, Z0_ref
)

# ============================================================
# 4. BUILD DEVICE AND TITLE INFORMATION
# ============================================================

# Use the original N_supercells directly; this sweep has no N override.
istar_sweep_device_length_mm = N_supercells * N_stubs_sc * p_cell * 1e3
istar_sweep_total_cells = N_supercells * N_stubs_sc

if "f_stop_s21" in globals():
    istar_sweep_stopband_text = f", fstop ~ {f_stop_s21/1e9:.3f} GHz"
else:
    istar_sweep_stopband_text = ""

# Build the common context shown above every Istar-sweep plot.
istar_sweep_title_common = (
    f"fpump = {ISTAR_SWEEP_PUMP_GHZ:.2f} GHz, "
    f"Ppump = {ISTAR_SWEEP_FIXED_PUMP_POWER_DBM:.1f} dBm, "
    f"Psignal = {ISTAR_SWEEP_SIGNAL_POWER_DBM:.1f} dBm, "
    f"N = {N_supercells}, "
    f"L = {istar_sweep_device_length_mm:.1f} mm, model = {GAIN_MODEL}"
    f"{istar_sweep_stopband_text}"
)

# Report all inputs before starting the expensive gain solves.
print("\n========== BASIC 4WM ISTAR SWEEP INPUTS ==========")
print(f"Istar values = {istar_sweep_values_ma} mA")
print(f"Fixed pump frequency = {ISTAR_SWEEP_PUMP_GHZ:.3f} GHz")
print(f"Fixed pump power = {ISTAR_SWEEP_FIXED_PUMP_POWER_DBM:.2f} dBm")
print(f"Fixed Ip0 = {istar_sweep_Ip0*1e3:.6f} mA")
print(f"Psignal = {ISTAR_SWEEP_SIGNAL_POWER_DBM:.2f} dBm")
print(f"Is0 = {istar_sweep_Is0*1e6:.6f} uA")
print(f"N_supercells = {N_supercells}")
print(f"N_total_cells = {istar_sweep_total_cells}")
print(f"Device length = {istar_sweep_device_length_mm:.3f} mm")
print(f"Gain model = {GAIN_MODEL}")
print(f"spatial_thin = {GAIN_SPATIAL_THIN}")
print(f"Spatial sample step = {GAIN_SPATIAL_THIN * p_cell * 1e6:.3f} um")

# ============================================================
# 5. CREATE THE SAME BASIC-4WM MODE SET AS THE OTHER SWEEPS
# ============================================================

istar_sweep_modes_4wm = ModeSet.basic(three_wave=False)
istar_sweep_terms_4wm = istar_sweep_modes_4wm.rwa_terms(power=3)
print(f"Number of basic-4WM RWA terms: {len(istar_sweep_terms_4wm)}")

# ============================================================
# 6. INITIALIZE RESULT STORAGE AND THE GAIN-CURVE FIGURE
# ============================================================

# Store one full curve and two scalar gain metrics per Istar value.
istar_sweep_results = {}
istar_sweep_peak_gain_db = []
istar_sweep_mean_gain_db = []
istar_sweep_ip_ratios = []
# Keep one analysis for stopband shading after the loop.
istar_sweep_reference_analysis = None

# Track whether diagnostic failed-point crosses are needed.
istar_sweep_has_failed_points = False

# Draw every Istar gain trace on this shared axis.
istar_sweep_fig, istar_sweep_ax = plt.subplots(figsize=(10, 6))

# ============================================================
# 7. SWEEP ISTAR AT THE FIXED PUMP POWER
# ============================================================

for istar_sweep_value_a in ISTAR_SWEEP_VALUES_A:
    # Convert to mA only for labels and reporting.
    istar_sweep_value_ma = istar_sweep_value_a * 1e3

    # Rebuild TWPA because Istar changes the nonlinear coefficient.
    # N, Istar, and Ip0 are passed directly, exactly as in the top TWPA.
    istar_sweep_twpa = TWPA(
        cells=cells,
        N=N_supercells,
        Istar=istar_sweep_value_a,
        Idc=Idc,
        Ip0=istar_sweep_Ip0,
        Z0_ref=Z0_ref,
    )

    # Build the analysis on the same shared frequency range as the workflow.
    istar_sweep_analysis = TWPAnalysis(
        twpa=istar_sweep_twpa,
        freqs=ANALYSIS_FREQUENCY_RANGE_GHZ,
        unit="GHz",
    )

    # Reject a pump that lies inside the configured stopband guard.
    if not pump_is_stopband_safe(istar_sweep_analysis, ISTAR_SWEEP_PUMP_GHZ):
        raise ValueError(
            f"Pump {ISTAR_SWEEP_PUMP_GHZ:.3f} GHz lies inside the configured stopband guard."
        )

    # All points share the linear device, so one analysis can shade the plot.
    if istar_sweep_reference_analysis is None:
        istar_sweep_reference_analysis = istar_sweep_analysis


    # Calculate the normalized drive without modifying either solver input.
    istar_sweep_ip_ratio = istar_sweep_Ip0 / istar_sweep_value_a
    istar_sweep_ip_ratios.append(istar_sweep_ip_ratio)


    # Report the current point before launching the gain solve.
    print("\n==================================================")
    print(f"RUNNING BASIC 4WM ISTAR = {istar_sweep_value_ma:.3f} mA")
    print(f"Fixed Ip0 = {istar_sweep_Ip0*1e3:.6f} mA")
    print(f"Ip0/Istar = {istar_sweep_ip_ratio:.6f}")
    print(f"Is0/Ip0 = {istar_sweep_Is0/istar_sweep_Ip0:.6e}")


    # Warn if this Istar makes the normalized pump unusually large.
    if istar_sweep_ip_ratio > PUMP_CURRENT_WARNING_RATIO:
        print(
            f"WARNING: Ip0/Istar > {PUMP_CURRENT_WARNING_RATIO:.2f}. "
            "This may be outside the reliable perturbative KI/CME regime."
        )


    # Use the shared helper so the Bloch configuration matches every block.
    istar_sweep_curve = run_gain_curve(
        istar_sweep_analysis,
        istar_sweep_Is0,
        float(ISTAR_SWEEP_PUMP_GHZ),
        Ip0_value=istar_sweep_Ip0,
        modes=istar_sweep_modes_4wm,
    )

    # Store the converged/masked gain returned by run_gain_curve.
    istar_sweep_gain_db = istar_sweep_curve["gain_db"]
    istar_sweep_results[istar_sweep_value_a] = istar_sweep_curve

    # Calculate the same peak and mean metrics as the pump-power sweep.
    istar_sweep_peak = np.nanmax(istar_sweep_gain_db)
    istar_sweep_mean = np.nanmean(istar_sweep_gain_db)
    istar_sweep_peak_gain_db.append(istar_sweep_peak)
    istar_sweep_mean_gain_db.append(istar_sweep_mean)


    # Add this trace and its failed-point diagnostics to the gain figure.
    _, istar_sweep_failed = plot_gain_curve(
        istar_sweep_ax,
        istar_sweep_curve,
        label=(
            f"Istar = {istar_sweep_value_ma:.2f} mA, "
            f"Ip0/Istar = {istar_sweep_ip_ratio:.3f}"
        ),
    )
    istar_sweep_has_failed_points |= bool(np.any(istar_sweep_failed))

    # Print the same convergence report as the other gain blocks.
    print_gain_diagnostics(
        istar_sweep_curve, f"Istar {istar_sweep_value_ma:.2f} mA"
    )
    print(f"Peak gain = {istar_sweep_peak:.3f} dB")
    print(f"Mean gain = {istar_sweep_mean:.3f} dB")

# ============================================================
# 8. PLOT GAIN VERSUS SIGNAL FREQUENCY FOR EVERY ISTAR
# ============================================================

# Shade the stopband and the corresponding idler-mapped region.
shade_stopbands(
    istar_sweep_ax,
    istar_sweep_reference_analysis,
    pump_ghz=ISTAR_SWEEP_PUMP_GHZ,
)

# Add a legend entry when any finite points failed convergence.
if istar_sweep_has_failed_points:
    istar_sweep_ax.scatter(
        [], [], marker="x", color="black", s=25,
        label="finite but not converged (diagnostic only)",
    )

# Apply the same labels, limits, grid, and title style as the pump sweep.
istar_sweep_ax.set_xlabel("Signal frequency [GHz]")
istar_sweep_ax.set_ylabel("Signal gain S21 [dB]")
istar_sweep_ax.set_title(
    "Basic 4WM gain vs signal frequency for Istar sweep\n"
    + istar_sweep_title_common
    + f", thin = {GAIN_SPATIAL_THIN}"
)
istar_sweep_ax.set_xlim(GAIN_SIGNAL_MIN_GHZ, GAIN_SIGNAL_MAX_GHZ)
istar_sweep_ax.grid(True, alpha=0.3)
istar_sweep_ax.legend(title="Istar")
istar_sweep_fig.tight_layout()
plt.show()

# ============================================================
# 9. PREPARE SCALAR RESULTS FOR THE TWO SUMMARY PLOTS
# ============================================================

istar_sweep_peak_gain_db = np.asarray(istar_sweep_peak_gain_db)
istar_sweep_mean_gain_db = np.asarray(istar_sweep_mean_gain_db)
istar_sweep_ip_ratios = np.asarray(istar_sweep_ip_ratios)

# All Istar traces share one signal-frequency grid.
istar_curve_columns = [
    next(iter(istar_sweep_results.values()))["signal_freqs"]
]
istar_curve_header = ["signal_frequency_ghz"]
for istar_value_a in ISTAR_SWEEP_VALUES_A:
    istar_curve_columns.append(
        istar_sweep_results[istar_value_a]["gain_db"]
    )
    istar_curve_header.append(
        f"gain_db_Istar_{istar_value_a * 1e3:.3f}_mA"
    )
combined = np.column_stack(istar_curve_columns)
istar_curves_filename = (
    f"gain_Istar_sweep_fpump{ISTAR_SWEEP_PUMP_GHZ:.3f}GHz_"
    f"Ppump{ISTAR_SWEEP_FIXED_PUMP_POWER_DBM:.1f}dBm_"
    f"Psignal{ISTAR_SWEEP_SIGNAL_POWER_DBM:.1f}dBm_"
    f"N{N_supercells}_curves.csv"
)
# np.savetxt(f"C:\\Users\\Lamon\\TWPA_solver\\{istar_curves_filename}", combined, delimiter=",", header=",".join(istar_curve_header), comments="", fmt="%g")
# np.savetxt(f"G:\\My Drive\\twpasolver_data\\{istar_curves_filename}", combined, delimiter=",", header=",".join(istar_curve_header), comments="", fmt="%g")

# One summary table covers the Istar and Ip0/Istar plots.
combined = np.column_stack((
    istar_sweep_values_for_checks, istar_sweep_values_ma,
    np.full_like(istar_sweep_values_ma, istar_sweep_Ip0 * 1e3),
    istar_sweep_ip_ratios, istar_sweep_peak_gain_db,
    istar_sweep_mean_gain_db,
))
istar_summary_filename = (
    f"gain_Istar_sweep_fpump{ISTAR_SWEEP_PUMP_GHZ:.3f}GHz_"
    f"Ppump{ISTAR_SWEEP_FIXED_PUMP_POWER_DBM:.1f}dBm_"
    f"Psignal{ISTAR_SWEEP_SIGNAL_POWER_DBM:.1f}dBm_"
    f"N{N_supercells}_summary.csv"
)
istar_summary_header = (
    "Istar_A,Istar_mA,Ip0_mA,Ip0_over_Istar,"
    "peak_gain_db,mean_gain_db"
)
# np.savetxt(f"C:\\Users\\Lamon\\TWPA_solver\\{istar_summary_filename}", combined, delimiter=",", header=istar_summary_header, comments="", fmt="%g")
# np.savetxt(f"G:\\My Drive\\twpasolver_data\\{istar_summary_filename}", combined, delimiter=",", header=istar_summary_header, comments="", fmt="%g")

# ============================================================
# 10. PLOT PEAK AND MEAN GAIN VERSUS ISTAR
# ============================================================
plt.figure(figsize=(8, 5))
plt.plot(
    istar_sweep_values_ma, istar_sweep_peak_gain_db, "o-", label="Peak gain"
)
plt.plot(
    istar_sweep_values_ma, istar_sweep_mean_gain_db, "s--", label="Mean gain"
)
plt.xlabel("Istar [mA]")
plt.ylabel("Gain [dB]")
plt.title("Basic 4WM gain vs Istar\n" + istar_sweep_title_common)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# ============================================================
# 11. PLOT PEAK AND MEAN GAIN VERSUS NORMALIZED PUMP CURRENT
# ============================================================
plt.figure(figsize=(8, 5))
plt.plot(
    istar_sweep_ip_ratios, istar_sweep_peak_gain_db, "o-", label="Peak gain"
)
plt.plot(
    istar_sweep_ip_ratios, istar_sweep_mean_gain_db, "s--", label="Mean gain"
)
plt.xlabel("Ip0 / Istar")
plt.ylabel("Gain [dB]")
plt.title(
    "Basic 4WM gain vs normalized pump current\n"
    + istar_sweep_title_common
)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# ============================================================
# 12. PRINT THE ISTAR-SWEEP SUMMARY TABLE
# ============================================================

print("\n========== BASIC 4WM ISTAR SWEEP SUMMARY ==========")
print(istar_sweep_title_common)
print(f"spatial_thin = {GAIN_SPATIAL_THIN}")
print(f"Spatial sample step = {GAIN_SPATIAL_THIN * p_cell * 1e6:.3f} um")
print()

for (
    istar_sweep_summary_a,
    istar_sweep_ratio_summary,
    istar_sweep_peak_summary,
    istar_sweep_mean_summary,
) in zip(
    ISTAR_SWEEP_VALUES_A,
    istar_sweep_ip_ratios,
    istar_sweep_peak_gain_db,
    istar_sweep_mean_gain_db,
):
    # Convert Istar to mA only for this printed row.
    istar_sweep_summary_ma = istar_sweep_summary_a * 1e3

    print(
        f"Istar = {istar_sweep_summary_ma:7.3f} mA | "
        f"Ppump = {ISTAR_SWEEP_FIXED_PUMP_POWER_DBM:6.1f} dBm | "
        f"Ip0 = {istar_sweep_Ip0*1e3:9.5f} mA | "
        f"Ip0/Istar = {istar_sweep_ratio_summary:8.5f} | "
        f"Is0/Ip0 = {istar_sweep_Is0/istar_sweep_Ip0:10.3e} | "
        f"Peak gain = {istar_sweep_peak_summary:8.3f} dB | "
        f"Mean gain = {istar_sweep_mean_summary:8.3f} dB"
    )
```

```python
# ============================================================
# DEEP DIAGNOSTIC: BETA / DELTA-BETA CONVENTIONS
# ============================================================
#
# Goal:
#   Compare the quantities actually used/stored by TWPAnalysis against
#   physical beta = 2*pi*f/v and against pm["delta"].
#
# This should reveal whether the scale issue is only a plotting convention
# or whether there is a real unit mismatch.

import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Settings are defined in the top workflow-settings section.
# -----------------------------

# -----------------------------
# Pull TWPAnalysis base data
# -----------------------------

freqs_for_k = np.asarray(analysis.freqs, dtype=float).flatten()
k_raw = np.real(np.asarray(analysis.linear.k)).flatten()

# Robust frequency unit check
if np.nanmax(freqs_for_k) > 1e6:
    freqs_for_k_ghz = freqs_for_k / 1e9
else:
    freqs_for_k_ghz = freqs_for_k

def k_cell_of_f_ghz(f_ghz):
    """
    Interpolate stored TWPAnalysis k.
    Based on earlier checks, this appears to be phase advance per cell,
    i.e. beta*p_cell, not beta in 1/m.
    """
    return np.interp(
        f_ghz,
        freqs_for_k_ghz,
        k_raw,
        left=np.nan,
        right=np.nan,
    )

def beta_from_analysis_per_m(f_ghz):
    """
    Convert stored k_cell to physical beta [1/m].
    """
    return k_cell_of_f_ghz(f_ghz) / p_cell

def beta_from_avg_velocity_per_m(f_ghz):
    """
    Simple average-line estimate beta = 2*pi*f/v_avg.
    Uses v_avg from your notebook.
    """
    return 2.0 * np.pi * f_ghz * 1e9 / v_avg

print("\n========== BETA CONVENTION CHECK ==========")
print(f"p_cell = {p_cell:.6e} m")
print(f"1/p_cell = {1/p_cell:.6e} 1/m")
print(f"v_avg = {v_avg:.6e} m/s")
print()
print("Frequency | k_analysis raw | beta_analysis=k/p | beta_2pi_f/v | beta ratio")
print("   [GHz]  |   [rad/cell]   |      [1/m]        |     [1/m]    | beta_analysis / beta_avg")
print("-" * 98)

for f in [4.0, 8.0, PHASE_DIAGNOSTIC_PUMP_GHZ, 12.0, f_stop_s21/1e9]:
    k_cell = k_cell_of_f_ghz(f)
    beta_analysis = beta_from_analysis_per_m(f)
    beta_avg = beta_from_avg_velocity_per_m(f)
    print(
        f"{f:8.3f} | "
        f"{k_cell:14.6e} | "
        f"{beta_analysis:16.6e} | "
        f"{beta_avg:12.6e} | "
        f"{beta_analysis/beta_avg:12.6e}"
    )

# -----------------------------
# Manual delta-beta comparisons at selected signal freqs
# -----------------------------

print("\n========== MANUAL DELTA-BETA CHECK ==========")
print(f"Pump frequency = {PHASE_DIAGNOSTIC_PUMP_GHZ:.6f} GHz")
print("Definition: delta_k_cell = k_s + k_i - 2*k_p")
print("            delta_beta_m = delta_k_cell / p_cell")
print()

print("fs [GHz] | fi [GHz] | delta_k_cell [rad/cell] | delta_beta [1/m] | delta_beta*L [rad]")
print("-" * 100)

device_length_m = N_supercells * N_stubs_sc * p_cell

for f_s in BETA_CHECK_SIGNAL_FREQS_GHZ:
    f_i = 2.0 * PHASE_DIAGNOSTIC_PUMP_GHZ - f_s

    k_s = k_cell_of_f_ghz(f_s)
    k_i = k_cell_of_f_ghz(f_i)
    k_p = k_cell_of_f_ghz(PHASE_DIAGNOSTIC_PUMP_GHZ)

    delta_k_cell = k_s + k_i - 2.0 * k_p
    delta_beta_m = delta_k_cell / p_cell

    print(
        f"{f_s:8.3f} | "
        f"{f_i:8.3f} | "
        f"{delta_k_cell:24.6e} | "
        f"{delta_beta_m:16.6e} | "
        f"{delta_beta_m*device_length_m:16.6e}"
    )

# -----------------------------
# Official pm["delta"] comparison
# -----------------------------
#
# The documented phase_matching() function returns pm["delta"].
# That is the safest quantity to compare to the solver's own phase-matching plot.

print("\n========== OFFICIAL phase_matching() RESULT CHECK ==========")
# print("pm keys:", list(pm.keys()))

if pm.delta is not None:
    delta_pm = np.asarray(pm.delta)
    pm_signal_freqs = np.asarray(pm.signal_freqs)
    pm_pump_freqs = np.asarray(pm.pump_freqs)

    # Find the nearest pump column to the configured diagnostic pump.
    pump_idx = np.nanargmin(
        np.abs(pm_pump_freqs - PHASE_DIAGNOSTIC_PUMP_GHZ)
    )
    pump_nearest = pm_pump_freqs[pump_idx]

    delta_vs_signal = delta_pm[:, pump_idx]

    combined = np.column_stack((pm_signal_freqs, delta_vs_signal))
    normalized_phase_filename = (
        f"phase_matching_normalized_slice_fpump{pump_nearest:.3f}GHz_"
        f"Ip0_Istar{Ip0/Istar:.6f}_N{N_supercells}.csv"
    )
    # np.savetxt(f"C:\\Users\\Lamon\\TWPA_solver\\{normalized_phase_filename}", combined, delimiter=",", header="signal_frequency_ghz,delta_k_over_ks", comments="", fmt="%g")
    # np.savetxt(f"G:\\My Drive\\twpasolver_data\\{normalized_phase_filename}", combined, delimiter=",", header="signal_frequency_ghz,delta_k_over_ks", comments="", fmt="%g")

    print(f"pm['delta'] shape = {delta_pm.shape}")
    print(f"nearest pump in pm grid = {pump_nearest:.6f} GHz")
    print(f"pm delta min/max at this pump = {np.nanmin(delta_vs_signal):.6e}, {np.nanmax(delta_vs_signal):.6e}")

    # Plot official pm delta for the selected pump
    plt.figure(figsize=(9, 5))
    plt.plot(pm_signal_freqs, delta_vs_signal, lw=2, label=r"official pm['delta']")
    plt.axhline(0, ls="--", lw=1)

    plt.xlabel("Signal frequency [GHz]")
    plt.ylabel(r"Normalized mismatch $\Delta k/k_s$")
    plt.title(
        "Official normalized twpasolver phase-matching result\n"
        f"nearest pump = {pump_nearest:.3f} GHz, "
        f"Ppump = {MAIN_PUMP_POWER_DBM:.1f} dBm displayed, "
        f"N = {N_supercells}, L = {device_length_mm:.2f} mm"
    )
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

else:
    print("No 'delta' key found in pm. phase_matching result structure changed.")
```

```python
# ============================================================
# LENGTH SWEEP: BASIC 4WM GAIN VS SIGNAL FREQUENCY
# ============================================================
#
# Assumes the following already exist from the main setup block:
#   cells
#   TWPA, TWPAnalysis, ModeArrayFactory
#   dBm_to_Ipeak
#   Istar, Idc, Z0_ref
#   N_stubs_sc, p_cell
#   f_stop_s21, f_stop_est   optional but used in titles if present
#
# Latest twpasolver assumed:
#   ModeArrayFactory.create_basic(..., three_wave=False)
#
# This block rebuilds TWPA/TWPAnalysis for each device length.
# That is necessary because N_supercells is stored inside TWPA(..., N=...).

import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Settings are defined in the top workflow-settings section.
# -----------------------------

Ip0_length_sweep = dBm_to_Ipeak(LENGTH_SWEEP_PUMP_POWER_DBM, Z0_ref)
Is0_length_sweep = dBm_to_Ipeak(LENGTH_SWEEP_SIGNAL_POWER_DBM, Z0_ref)

mode_array_config_name = "basic_4wm"

# Optional title text
if "f_stop_s21" in globals():
    stopband_text = f", fstop ≈ {f_stop_s21/1e9:.3f} GHz"
else:
    stopband_text = ""

print("\n========== BASIC 4WM LENGTH SWEEP INPUTS ==========")
print(f"Pump frequency = {LENGTH_SWEEP_PUMP_GHZ:.3f} GHz")
print(f"Ppump = {LENGTH_SWEEP_PUMP_POWER_DBM:.2f} dBm")
print(f"Ip0 = {Ip0_length_sweep*1e3:.6f} mA")
print(f"Ip0/Istar = {Ip0_length_sweep/Istar:.6f}")

print(f"\nPsignal = {LENGTH_SWEEP_SIGNAL_POWER_DBM:.2f} dBm")
print(f"Is0 = {Is0_length_sweep*1e6:.6f} uA")
print(f"Is0/Istar = {Is0_length_sweep/Istar:.6e}")
print(f"Is0/Ip0 = {Is0_length_sweep/Ip0_length_sweep:.6e}")

print(f"\nGain model = {GAIN_MODEL}")
print(f"spatial_thin = {GAIN_SPATIAL_THIN} cell(s)")
print(f"Spatial sample step = {GAIN_SPATIAL_THIN * p_cell * 1e6:.3f} um")

if Ip0_length_sweep / Istar > PUMP_CURRENT_WARNING_RATIO:
    print(
        f"\nWARNING: Ip0/Istar > {PUMP_CURRENT_WARNING_RATIO:.2f}. "
        "This may be outside the reliable perturbative KI/CME regime."
    )

# -----------------------------
# Storage
# -----------------------------

gain_by_N = {}
peak_gain_by_N = []
mean_gain_by_N = []
length_mm_by_N = []
N_total_cells_by_N = []

# -----------------------------
# Sweep device length
# -----------------------------

for N_this in LENGTH_SWEEP_N_SUPERCELLS:

    N_total_cells = int(N_this) * N_stubs_sc
    length_mm = N_total_cells * p_cell * 1e3
    expected_samples = int(np.ceil(N_total_cells / GAIN_SPATIAL_THIN))

    print("\n==================================================")
    print(f"RUNNING BASIC 4WM LENGTH SWEEP: N_supercells = {N_this}")
    print(f"N_total_cells = {N_total_cells}")
    print(f"Device length = {length_mm:.3f} mm")
    print(f"spatial_thin = {GAIN_SPATIAL_THIN}")
    print(f"Expected spatial samples ≈ {expected_samples}")
    print(f"Spatial step = {GAIN_SPATIAL_THIN * p_cell * 1e6:.3f} um")

    # Rebuild TWPA with same cells, different number of supercells,
    # and pump current derived from LENGTH_SWEEP_PUMP_POWER_DBM.
    twpa_this = TWPA(
        cells=cells,
        N=int(N_this),
        Istar=Istar,
        Idc=Idc,
        Ip0=float(Ip0_length_sweep),
        Z0_ref=Z0_ref,
    )

    analysis_this = TWPAnalysis(
        twpa=twpa_this,
        freqs=ANALYSIS_FREQUENCY_RANGE_GHZ,
        unit= "GHz"
    )



    # Latest-version basic 4WM mode array:
    #   p = pump
    #   s = signal
    #   i = 2p - s
    from twpasolver import ModeSet

    modes_best = ModeSet.basic(
        three_wave=False,  # fi = 2*fp - fs
    )

    # Optional: print the mode expressions once for the first sweep point
    if N_this == LENGTH_SWEEP_N_SUPERCELLS[0]:
        print("\n========== BASIC 4WM MODE ARRAY CHECK ==========")
        # modes_best.print_symbolic_expressions()

        print("\n========== BASIC 4WM RWA TERMS ==========")
        terms_4wm = modes_best.rwa_terms(power=3)
        print(f"Number of 4WM RWA terms found: {len(terms_4wm)}")
        for term in terms_4wm:
            print(term)

    if not pump_is_stopband_safe(analysis_this, LENGTH_SWEEP_PUMP_GHZ):
        raise ValueError(
            f"Pump {LENGTH_SWEEP_PUMP_GHZ:.3f} GHz lies inside the configured stopband guard."
        )

    curve = run_gain_curve(
        analysis_this,
        Is0_length_sweep,
        LENGTH_SWEEP_PUMP_GHZ,
        Ip0_value=Ip0_length_sweep,
        modes=modes_best,
    )
    gain_db = curve["gain_db"]
    gain_by_N[int(N_this)] = curve

    peak_gain = np.nanmax(gain_db)
    mean_gain = np.nanmean(gain_db)

    peak_gain_by_N.append(peak_gain)
    mean_gain_by_N.append(mean_gain)
    length_mm_by_N.append(length_mm)
    N_total_cells_by_N.append(N_total_cells)

    print(f"Peak gain = {peak_gain:.3f} dB")
    print(f"Mean gain = {mean_gain:.3f} dB")
    print_gain_diagnostics(curve, f"N={int(N_this)}")


# Convert lists to arrays
peak_gain_by_N = np.asarray(peak_gain_by_N)
mean_gain_by_N = np.asarray(mean_gain_by_N)
length_mm_by_N = np.asarray(length_mm_by_N)
N_total_cells_by_N = np.asarray(N_total_cells_by_N)

# All device-length traces share one signal-frequency grid.
length_curve_columns = [
    next(iter(gain_by_N.values()))["signal_freqs"]
]
length_curve_header = ["signal_frequency_ghz"]
for N_this, length_mm in zip(LENGTH_SWEEP_N_SUPERCELLS, length_mm_by_N):
    length_curve_columns.append(gain_by_N[int(N_this)]["gain_db"])
    length_curve_header.append(
        f"gain_db_N{int(N_this)}_length_{length_mm:.3f}_mm"
    )
combined = np.column_stack(length_curve_columns)
length_curves_filename = (
    f"gain_device_length_sweep_fpump{LENGTH_SWEEP_PUMP_GHZ:.3f}GHz_"
    f"Ppump{LENGTH_SWEEP_PUMP_POWER_DBM:.1f}dBm_"
    f"Psignal{LENGTH_SWEEP_SIGNAL_POWER_DBM:.1f}dBm_"
    f"Istar{Istar * 1e3:.3f}mA_curves.csv"
)
# np.savetxt(f"C:\\Users\\Lamon\\TWPA_solver\\{length_curves_filename}", combined, delimiter=",", header=",".join(length_curve_header), comments="", fmt="%g")
# np.savetxt(f"G:\\My Drive\\twpasolver_data\\{length_curves_filename}", combined, delimiter=",", header=",".join(length_curve_header), comments="", fmt="%g")

# One summary table covers the physical-length and supercell-count plots.
combined = np.column_stack((
    LENGTH_SWEEP_N_SUPERCELLS, N_total_cells_by_N, length_mm_by_N,
    peak_gain_by_N, mean_gain_by_N,
))
length_summary_filename = (
    f"gain_device_length_sweep_fpump{LENGTH_SWEEP_PUMP_GHZ:.3f}GHz_"
    f"Ppump{LENGTH_SWEEP_PUMP_POWER_DBM:.1f}dBm_"
    f"Psignal{LENGTH_SWEEP_SIGNAL_POWER_DBM:.1f}dBm_"
    f"Istar{Istar * 1e3:.3f}mA_summary.csv"
)
length_summary_header = (
    "N_supercells,N_total_cells,length_mm,peak_gain_db,mean_gain_db"
)
# np.savetxt(f"C:\\Users\\Lamon\\TWPA_solver\\{length_summary_filename}", combined, delimiter=",", header=length_summary_header, comments="", fmt="%g")
# np.savetxt(f"G:\\My Drive\\twpasolver_data\\{length_summary_filename}", combined, delimiter=",", header=length_summary_header, comments="", fmt="%g")

# ============================================================
# Plot 1: Gain curves vs signal frequency for all lengths
# ============================================================

fig, ax = plt.subplots(figsize=(10, 6))
has_failed_points = False

for N_this, length_mm in zip(LENGTH_SWEEP_N_SUPERCELLS, length_mm_by_N):
    _, failed = plot_gain_curve(
        ax, gain_by_N[int(N_this)],
        label=f"N={int(N_this)}, L={length_mm:.1f} mm",
    )
    has_failed_points |= bool(np.any(failed))

shade_stopbands(ax, analysis_this, pump_ghz=LENGTH_SWEEP_PUMP_GHZ)
if has_failed_points:
    ax.scatter(
        [], [], marker="x", color="black", s=25,
        label="finite but not converged (diagnostic only)",
    )
ax.set_xlabel("Signal frequency [GHz]")
ax.set_ylabel("Signal gain S21 [dB]")
ax.set_title(
    "Basic 4WM gain vs signal frequency for length sweep\n"
    f"fpump = {LENGTH_SWEEP_PUMP_GHZ:.2f} GHz, "
    f"Ppump = {LENGTH_SWEEP_PUMP_POWER_DBM:.1f} dBm, "
    f"Psignal = {LENGTH_SWEEP_SIGNAL_POWER_DBM:.1f} dBm, "
    f"model = {GAIN_MODEL}"
    "\n"
    f"Ip0/I* = {Ip0_length_sweep/Istar:.3f}, "
    f"thin = {GAIN_SPATIAL_THIN}{stopband_text}"
)
ax.set_xlim(GAIN_SIGNAL_MIN_GHZ, GAIN_SIGNAL_MAX_GHZ)
ax.grid(True, alpha=0.3)
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()


# ============================================================
# Plot 2: Peak / mean gain vs physical length
# ============================================================

plt.figure(figsize=(8, 5))
plt.plot(length_mm_by_N, peak_gain_by_N, "o-", label="Peak gain")
plt.plot(length_mm_by_N, mean_gain_by_N, "s--", label="Mean gain")
plt.xlabel("Physical length [mm]")
plt.ylabel("Gain [dB]")
plt.title(
    "Basic 4WM gain vs device length\n"
    f"fpump = {LENGTH_SWEEP_PUMP_GHZ:.2f} GHz, "
    f"Ppump = {LENGTH_SWEEP_PUMP_POWER_DBM:.1f} dBm, "
    f"Psignal = {LENGTH_SWEEP_SIGNAL_POWER_DBM:.1f} dBm, "
    f"model = {GAIN_MODEL}"
    "\n"
    f"Ip0/I* = {Ip0_length_sweep/Istar:.3f}, "
    f"thin = {GAIN_SPATIAL_THIN}{stopband_text}"
)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# Plot 3: Peak gain vs number of supercells
# ============================================================

plt.figure(figsize=(8, 5))
plt.plot(LENGTH_SWEEP_N_SUPERCELLS, peak_gain_by_N, "o-", label="Peak gain")
plt.plot(LENGTH_SWEEP_N_SUPERCELLS, mean_gain_by_N, "s--", label="Mean gain")
plt.xlabel("Number of supercells")
plt.ylabel("Gain [dB]")
plt.title(
    "Basic 4WM gain vs number of supercells\n"
    f"fpump = {LENGTH_SWEEP_PUMP_GHZ:.2f} GHz, "
    f"Ppump = {LENGTH_SWEEP_PUMP_POWER_DBM:.1f} dBm, model = {GAIN_MODEL}"
)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# Print summary table
# ============================================================

print("\n========== BASIC 4WM LENGTH SWEEP SUMMARY ==========")
print(f"Pump frequency = {LENGTH_SWEEP_PUMP_GHZ:.3f} GHz")
print(f"Gain model = {GAIN_MODEL}")
print(f"spatial_thin = {GAIN_SPATIAL_THIN}")
print(f"Ppump = {LENGTH_SWEEP_PUMP_POWER_DBM:.2f} dBm | Ip0 = {Ip0_length_sweep*1e3:.6f} mA | Ip0/Istar = {Ip0_length_sweep/Istar:.6f}")
print(f"Psignal = {LENGTH_SWEEP_SIGNAL_POWER_DBM:.2f} dBm | Is0 = {Is0_length_sweep*1e6:.6f} uA | Is0/Ip0 = {Is0_length_sweep/Ip0_length_sweep:.6e}")

for N_this, N_total_cells, length_mm, peak_gain, mean_gain in zip(
    LENGTH_SWEEP_N_SUPERCELLS,
    N_total_cells_by_N,
    length_mm_by_N,
    peak_gain_by_N,
    mean_gain_by_N,
):
    print(
        f"N = {int(N_this):4d} | "
        f"cells = {int(N_total_cells):6d} | "
        f"L = {length_mm:8.3f} mm | "
        f"Peak gain = {peak_gain:8.3f} dB | "
        f"Mean gain = {mean_gain:8.3f} dB"
    )
```
